```python
# ============================================================
# Segmentación no supervisada de imagen con RAPIDS
# Basado en:
# - Clustering no supervisado (KMeans, inercia, silueta)  :contentReference[oaicite:13]{index=13}
# - Segmentación de imagen con info espacial tipo "tanjiro"  :contentReference[oaicite:14]{index=14}
# - Reducción de dimensión (PCA / UMAP) 2D  :contentReference[oaicite:15]{index=15}
# ============================================================

import numpy as np
import cupy as cp
import cudf
import matplotlib.pyplot as plt
import cv2

from cuml.cluster import KMeans
from cuml.decomposition import PCA
from cuml.manifold import UMAP

from sklearn.metrics import silhouette_score  # CPU fallback

# -----------------------------
# (1) Cargar imagen y recortar
# -----------------------------
IMAGE_PATH = "/kaggle/input/tanjiroo/tanjiro.jpeg"

img_bgr = cv2.imread(IMAGE_PATH)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

# recorte tipo tanjiro: image = image[50:175,50:175,:]  :contentReference[oaicite:16]{index=16}
img_rgb = img_rgb[50:175, 50:175, :]
h, w, ch = img_rgb.shape
print("Imagen recortada:", img_rgb.shape)

# -----------------------------
# (2) Construir matriz de pixeles [R,G,B,x,y]
# -----------------------------
ys, xs = np.meshgrid(np.arange(h), np.arange(w), indexing='ij')

R = img_rgb[:, :, 0].reshape(-1)
G = img_rgb[:, :, 1].reshape(-1)
B = img_rgb[:, :, 2].reshape(-1)
Xc = xs.reshape(-1)  # columna (x)
Xr = ys.reshape(-1)  # fila (y)

# Stack: color + posición espacial
X_full = np.stack([R, G, B, Xc, Xr], axis=1).astype(np.float32)

# Escalamiento simple como en tanjiro: MinMax-like manual
# color va de 0-255 -> escalar a [0,1]
X_full[:,0:3] /= 255.0
# posicional la llevamos a [0,1] dividiendo por tamaño
X_full[:,3] /= float(w)
X_full[:,4] /= float(h)

print("Matriz de características original RGB+XY:", X_full.shape)

# Pasar a GPU
X_full_cu = cp.asarray(X_full)                # cupy (N,5)
X_full_gdf = cudf.DataFrame(X_full, columns=["R","G","B","x","y"])  # cudf


# ============================================================
# Funciones auxiliares
# ============================================================

def compute_inertia_gpu(X_cu, labels_cu, centers_cu):
    # X_cu: cupy (N,d)
    # labels_cu: cupy (N,)
    # centers_cu: cupy (k,d)
    assigned = centers_cu[labels_cu]  # (N,d)
    diff = X_cu - assigned
    dist_sq = cp.sum(diff**2, axis=1)
    inertia_val = cp.sum(dist_sq)
    return float(inertia_val.get())

def compute_silhouette_cpu_sampled(X_np, labels_np, max_samples=20000):
    # silhouette_score de sklearn corre en CPU
    if len(np.unique(labels_np)) < 2:
        return np.nan
    if X_np.shape[0] > max_samples:
        idx = np.random.choice(X_np.shape[0], size=max_samples, replace=False)
        X_use = X_np[idx]
        y_use = labels_np[idx]
    else:
        X_use = X_np
        y_use = labels_np
    return silhouette_score(X_use, y_use, metric="euclidean")

def kmeans_and_eval(X_gdf, X_cu, k_list, h, w, original_rgb, tag_space):
    """
    - Ajusta KMeans (cuML) para cada k en k_list
    - Calcula inercia y silueta
    - Reconstruye segmentación visual
    """
    results = []

    # Para reconstrucción de imagen segmentada necesitamos los colores originales (R,G,B sin escalar)
    R0 = original_rgb[:,:,0].reshape(-1)
    G0 = original_rgb[:,:,1].reshape(-1)
    B0 = original_rgb[:,:,2].reshape(-1)

    for k in k_list:
        print(f"\n--- {tag_space} | k={k} ---")
        km = KMeans(n_clusters=k, max_iter=300, random_state=42)
        km.fit(X_gdf)

        # labels -> cupy
        labels_cu = km.labels_.values  # cudf Series -> cupy
        labels_np = cp.asnumpy(labels_cu)

        # centros -> cupy
        centers_cu = cp.asarray(km.cluster_centers_.values)  # cudf DF -> cupy

        # métrica de inercia (como en 7_ConglomeradosNosupervisado)  :contentReference[oaicite:17]{index=17}
        inertia_val = compute_inertia_gpu(X_cu, labels_cu, centers_cu)

        # métrica de silueta (coeficiente silueta)  :contentReference[oaicite:18]{index=18}
        X_host = cp.asnumpy(X_cu)
        sil_val = compute_silhouette_cpu_sampled(X_host, labels_np, max_samples=20000)

        print(f"Inercia = {inertia_val:.4f}")
        print(f"Silueta ≈ {sil_val:.4f}")

        # Reconstruir segmentación con color promedio por clúster (como en tanjiro)  :contentReference[oaicite:19]{index=19}
        seg_rgb = np.zeros((h*w,3), dtype=np.float32)
        for cluster_id in range(k):
            mask = labels_np == cluster_id
            if np.any(mask):
                seg_rgb[mask,0] = np.mean(R0[mask])
                seg_rgb[mask,1] = np.mean(G0[mask])
                seg_rgb[mask,2] = np.mean(B0[mask])

        seg_rgb = seg_rgb.reshape(h,w,3).astype(np.uint8)

        # Mostrar segmentación
        fig, ax = plt.subplots(1,2, figsize=(10,5))
        ax[0].imshow(original_rgb)
        ax[0].set_title("Original")
        ax[0].axis("off")

        ax[1].imshow(seg_rgb)
        ax[1].set_title(f"{tag_space} - K={k}")
        ax[1].axis("off")
        plt.tight_layout()
        plt.show()

        results.append({
            "espacio": tag_space,
            "k": k,
            "inercia": inertia_val,
            "silhouette": sil_val
        })
    return results


# ============================================================
# EJERCICIO 1:
# Clustering directo sobre RGB+XY
# ============================================================

k_values = [3,5,7,9]

res_base = kmeans_and_eval(
    X_gdf = X_full_gdf,
    X_cu  = X_full_cu,
    k_list = k_values,
    h = h,
    w = w,
    original_rgb = img_rgb,
    tag_space = "RGB+XY"
)


# ============================================================
# EJERCICIO 2:
# Reducción de dimensión (PCA y UMAP) -> luego clustering
# ============================================================

# ---- PCA 2D en GPU (cuml.decomposition.PCA)  :contentReference[oaicite:20]{index=20}
pca_model = PCA(n_components=2, random_state=42)
X_pca_gdf = pca_model.fit_transform(X_full_gdf)       # cudf (N,2)
X_pca_cu  = cp.asarray(X_pca_gdf.values)              # cupy (N,2)
print("PCA 2D:", X_pca_gdf.shape)

res_pca = kmeans_and_eval(
    X_gdf = X_pca_gdf,
    X_cu  = X_pca_cu,
    k_list = k_values,
    h = h,
    w = w,
    original_rgb = img_rgb,
    tag_space = "PCA-2D"
)

# ---- UMAP 2D en GPU (cuml.manifold.UMAP)  :contentReference[oaicite:21]{index=21}
umap_model = UMAP(n_neighbors=15, n_components=2, random_state=42)
# cuML UMAP acepta cupy/cudf; usamos cupy directo:
X_umap_cu = umap_model.fit_transform(X_full_cu)       # cupy (N,2)
# convertir a cuDF para KMeans
X_umap_np = cp.asnumpy(X_umap_cu)
X_umap_gdf = cudf.DataFrame(X_umap_np, columns=["UMAP1","UMAP2"])
print("UMAP 2D:", X_umap_gdf.shape)

res_umap = kmeans_and_eval(
    X_gdf = X_umap_gdf,
    X_cu  = X_umap_cu,
    k_list = k_values,
    h = h,
    w = w,
    original_rgb = img_rgb,
    tag_space = "UMAP-2D"
)


# ============================================================
# TABLA RESUMEN FINAL
# ============================================================

import pandas as pd
df_resultados = pd.DataFrame(res_base + res_pca + res_umap)
print(df_resultados)

# Gráfica comparativa de Silueta vs k
plt.figure(figsize=(7,5))
for espacio in df_resultados["espacio"].unique():
    sub = df_resultados[df_resultados["espacio"] == espacio]
    plt.plot(sub["k"], sub["silhouette"], marker='o', label=espacio)
plt.xlabel("k")
plt.ylabel("Silhouette promedio")
plt.title("Comparación Silueta por representación")
plt.legend()
plt.grid(alpha=0.4, linestyle="--")
plt.show()

# Gráfica comparativa de Inercia vs k
plt.figure(figsize=(7,5))
for espacio in df_resultados["espacio"].unique():
    sub = df_resultados[df_resultados["espacio"] == espacio]
    plt.plot(sub["k"], sub["inercia"], marker='s', label=espacio)
plt.xlabel("k")
plt.ylabel("Inercia total")
plt.title("Comparación Inercia por representación")
plt.legend()
plt.grid(alpha=0.4, linestyle="--")
plt.show()
```

# Ejercicio 1 

### Ejercicio 1 — PCA: modelo generativo vs. maximización de varianza

El **Análisis de Componentes Principales (PCA)** puede entenderse desde dos perspectivas: **generativa** y **de varianza**.

---

#### 🔹 Enfoque generativo

El PCA asume que los datos originales \$ \mathbf{X} \in \mathbb{R}^{N \times P} \$ provienen de un espacio latente de menor dimensión  
\$ \mathbf{Z} \in \mathbb{R}^{N \times M} \$, con \$ M < P $\.

El modelo busca una matriz de pesos \$ \mathbf{W} \in \mathbb{R}^{P \times M} \ que reconstruya los datos como:

$$
\mathbf{X} \approx \mathbf{Z}\mathbf{W}^T
$$

El problema de optimización se formula como:

$$
\mathbf{W}^* = \arg\min_{\mathbf{W}} \| \mathbf{X} - \mathbf{Z}\mathbf{W}^T \|_F^2
$$

bajo la restricción ortonormal:

$$
\mathbf{W}\mathbf{W}^T = \mathbf{I}
$$

En este sentido, PCA actúa como un **filtro o generador de datos limpios**, reconstruyendo una versión comprimida del conjunto original.

---

#### 🔹 Enfoque de preservación de varianza

Desde otra perspectiva, PCA busca una transformación lineal que proyecte los datos en los ejes donde **la varianza es máxima**.  
El problema equivalente se expresa como:

$$
\mathbf{W}^* = \arg\max_{\mathbf{W}} \operatorname{tr}(\mathbf{W}^T \mathbf{\Sigma} \mathbf{W})
$$

donde:

$$
\mathbf{\Sigma} = \frac{1}{N} \mathbf{X}^T \mathbf{X}
$$

es la matriz de covarianza.

La solución se obtiene eligiendo los **autovectores asociados a los mayores autovalores** de \$ \mathbf{\Sigma} \$.

---

#### 🔹 Conclusión

Ambos enfoques son **equivalentes**:
- El enfoque generativo busca **reconstruir bien los datos**.  
- El enfoque de varianza busca **preservar la información más importante**.

La diferencia está en su interpretación: uno **“genera”** (minimiza error), y el otro **“maximiza”** (la varianza).


# Ejercicio 2

### Ejercicio 2 — Comparativo entre PCA y Kernel PCA

El **PCA** lineal busca una proyección que maximice la varianza en el espacio original, mientras que el **Kernel PCA (KPCA)** amplía este concepto a **espacios no lineales** mediante el uso del *kernel trick*.

---

#### 🔹 PCA clásico

En el PCA tradicional, los datos se proyectan linealmente:

$$
\mathbf{Z} = \mathbf{X}\mathbf{W}
$$

donde \( \mathbf{W} \) contiene los vectores propios de la matriz de covarianza \( \mathbf{X}^T \mathbf{X} \).  
El objetivo es:

$$
\max_{\mathbf{W}} \operatorname{tr}(\mathbf{W}^T \mathbf{\Sigma} \mathbf{W}), \quad \text{s.a. } \mathbf{W}^T\mathbf{W} = \mathbf{I}
$$

PCA es **lineal**, y solo puede capturar relaciones rectas o globales entre las variables.

---

#### 🔹 Kernel PCA

El Kernel PCA aplica el mismo principio, pero en un **espacio de características no lineal** definido por una función de mapeo \( \phi(\cdot) \).  
No se calcula directamente \( \phi(x) \), sino una **matriz de kernel** \( \mathbf{K} \):

$$
K_{ij} = \kappa(\mathbf{x}_i, \mathbf{x}_j) = \langle \phi(\mathbf{x}_i), \phi(\mathbf{x}_j) \rangle
$$

Por ejemplo, con un kernel RBF:

$$
K_{ij} = \exp\left(-\frac{\|\mathbf{x}_i - \mathbf{x}_j\|^2}{2\sigma^2}\right)
$$

Luego se resuelve el problema de autovalores:

$$
\mathbf{K}\boldsymbol{\alpha} = \lambda \boldsymbol{\alpha}
$$

y las proyecciones se obtienen como:

$$
z_n = \sum_{i=1}^N \alpha_i \kappa(\mathbf{x}_n, \mathbf{x}_i)
$$

---

#### 🔹 Conclusión

- **PCA:** busca ejes lineales de máxima varianza en el espacio original.  
- **Kernel PCA:** busca ejes *no lineales* de máxima varianza en un espacio transformado por el kernel.

PCA sirve para datos aproximadamente planos; KPCA puede **“desenrollar”** estructuras complejas o curvas.


```python
# ====================================================
# Ejercicio 2 — Ejemplo práctico PCA vs. Kernel PCA
# ====================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA, KernelPCA
from sklearn.datasets import make_circles

# Generamos un dataset no lineal (dos círculos concéntricos)
X, y = make_circles(n_samples=400, factor=0.3, noise=0.05, random_state=42)

# Aplicamos PCA lineal
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Aplicamos Kernel PCA con kernel RBF (no lineal)
kpca = KernelPCA(n_components=2, kernel="rbf", gamma=15)
X_kpca = kpca.fit_transform(X)

# Visualizamos comparativamente
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.title("PCA lineal")
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis')
plt.xlabel("Componente 1"); plt.ylabel("Componente 2")

plt.subplot(1, 2, 2)
plt.title("Kernel PCA (RBF)")
plt.scatter(X_kpca[:, 0], X_kpca[:, 1], c=y, cmap='viridis')
plt.xlabel("Componente 1"); plt.ylabel("Componente 2")

plt.tight_layout()
plt.show()
```

# Ejercicio 3

### Ejercicio 3 — Eigenfaces y comparación con PCA

El algoritmo **Eigenfaces**, propuesto por Turk y Pentland (1991), se basa directamente en el método PCA aplicado a imágenes faciales.

---

#### 🔹 Fundamento del modelo

Cada imagen se representa como un vector en un espacio de alta dimensión (todos los píxeles concatenados).  
El PCA encuentra las direcciones principales de variación entre las imágenes, llamadas **eigenfaces**, que pueden verse como “caras base”.

Cada rostro puede representarse como combinación lineal de estas bases:

$$
\mathbf{x}_i \approx \bar{\mathbf{x}} + \sum_{k=1}^{M} a_{ik} \mathbf{u}_k
$$

donde:
- \$ \bar{\mathbf{x}} \$ es la imagen promedio,
- \$ \mathbf{u}_k \$ son los vectores propios (eigenfaces),
- \$ a_{ik} \$ son los coeficientes de proyección.

---

#### 🔹 Relación con PCA

El algoritmo de eigenfaces **no es distinto de PCA**, sino una **aplicación de PCA a datos visuales**:
- En PCA, los componentes principales representan direcciones de máxima varianza en cualquier conjunto de datos.
- En Eigenfaces, esos componentes tienen interpretación **visual**: cada eigenface resalta zonas típicas de variación (ojos, nariz, boca).

---

#### 🔹 Conclusión

El modelo Eigenfaces **reduce la dimensión** conservando la estructura visual esencial.  
Esto permite representar rostros o dígitos (como en MNIST) de manera compacta, facilitando tareas de clasificación o reconocimiento en un subespacio de menor dimensión.


```python
# ==========================================================
# Ejercicio 3 — Visualización Eigenfaces + PCA 2D con imágenes
# ==========================================================
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA
from matplotlib.offsetbox import OffsetImage, AnnotationBbox

# -----------------------------
# 1. Cargar conjunto de datos MNIST
# -----------------------------
mnist = fetch_openml('mnist_784', version=1, as_frame=False)
X = mnist.data[:1000].astype(np.float32)  # muestra pequeña por rendimiento
y = mnist.target[:1000].astype(int)

img_w, img_h = 28, 28

# -----------------------------
# 2. Aplicar PCA (Eigenfaces)
# -----------------------------
pca = PCA(n_components=0.95, svd_solver='full', random_state=42)
X_pca = pca.fit_transform(X)

print(f"Número de componentes retenidas: {pca.components_.shape[0]}")
print(f"Varianza explicada total: {np.sum(pca.explained_variance_ratio_):.2%}")

# -----------------------------
# 3. Mostrar las primeras 5 eigenfaces
# -----------------------------
fig, axes = plt.subplots(1, 5, figsize=(10, 2))
for i, ax in enumerate(axes):
    ax.imshow(pca.components_[i].reshape(img_w, img_h), cmap='gray')
    ax.set_title(f"Eigenface {i+1}")
    ax.axis("off")
plt.suptitle("Primeras 5 'Eigenfaces' (Componentes principales de PCA)")
plt.show()

# -----------------------------
# 4. Visualización PCA 2D con imágenes originales
# -----------------------------
Z = X_pca[:, :2]  # Tomamos las dos primeras componentes
plt.figure(figsize=(10, 8))
plt.title("Proyección PCA 2D con Imágenes Originales (MNIST)")

# Dibujar los puntos
plt.scatter(Z[:, 0], Z[:, 1], c=y, cmap='tab10', s=15, alpha=0.4)
plt.xlabel("Componente principal 1")
plt.ylabel("Componente principal 2")

# Agregar imágenes miniatura sobre los puntos (muestreo para no saturar)
for i in range(0, len(Z), 40):  # muestra 1 de cada 40 imágenes
    img = X[i].reshape(img_w, img_h)
    imagebox = OffsetImage(img, zoom=0.5, cmap='gray')
    ab = AnnotationBbox(imagebox, (Z[i, 0], Z[i, 1]), frameon=False)
    plt.gca().add_artist(ab)

plt.colorbar(label="Dígito")
plt.show()
```

```python
# ==========================================================
# Ejercicio 3 — Eigenfaces + comparación con métodos DR
# Versión robusta (tolera fallo de UMAP)
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
from sklearn.decomposition import PCA, KernelPCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Intentaremos importar umap, pero puede fallar más adelante
import umap

# -----------------------------
# 1. Cargar conjunto de datos MNIST
# -----------------------------
# Usamos 2000 muestras para que t-SNE no sea demasiado lento
mnist = fetch_openml('mnist_784', version=1, as_frame=False)

# Forzamos a float32 por estabilidad numérica y compatibilidad
X = mnist.data[:2000].astype(np.float32)
y = mnist.target[:2000].astype(int)

img_w, img_h = 28, 28

print("Shape X:", X.shape)
print("Shape y:", y.shape)

# -----------------------------
# 2. "Eigenfaces" (en MNIST serían "eigendígitos")
#    PCA para obtener componentes principales
# -----------------------------
# Retenemos ~95% de la varianza
pca = PCA(n_components=0.95, svd_solver='full', random_state=42)
X_pca_full = pca.fit_transform(X)

print("Shape X_pca_full:", X_pca_full.shape)
print("Número de componentes PCA retenidos:", pca.components_.shape[0])

# Mostrar las primeras 5 "eigenfaces"/"eigendígitos"
fig, axes = plt.subplots(1, 5, figsize=(10, 2))
for i, ax in enumerate(axes):
    ax.imshow(pca.components_[i].reshape(img_w, img_h), cmap='gray')
    ax.set_title(f"Eigen-{i+1}")
    ax.axis("off")
plt.suptitle("Primeras 5 'eigenfaces' (componentes principales de PCA)")
plt.show()

# -----------------------------
# 3. Visualización PCA 2D
#    Tomamos solo las dos primeras componentes principales
# -----------------------------
Z_pca2 = X_pca_full[:, :2]  # (2000, 2)

plt.figure(figsize=(8,6))
plt.scatter(Z_pca2[:, 0], Z_pca2[:, 1], c=y, cmap='tab10', s=10)
plt.title("Proyección PCA 2D (MNIST)")
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.colorbar(label="Dígito")
plt.show()

# -----------------------------
# 4. Otras reducciones de dimensión a 2D
#    - Kernel PCA (RBF)
#    - t-SNE
#    - UMAP (si funciona)
# -----------------------------

print("Generando Kernel PCA (RBF)...")
kpca = KernelPCA(n_components=2, kernel="rbf", gamma=0.01, random_state=42)
Z_kpca2 = kpca.fit_transform(X)  # (2000, 2)

print("Generando t-SNE...")
tsne = TSNE(n_components=2, perplexity=30, random_state=42)
Z_tsne2 = tsne.fit_transform(X)  # (2000, 2)

print("Intentando generar UMAP...")
Z_umap2 = None
umap_error = False
try:
    umap_model = umap.UMAP(
        n_components=2,
        n_neighbors=40,
        min_dist=0.3,
        random_state=42,
    )
    # En tu entorno, .fit_transform(X) revienta con ensure_all_finite
    # Intento directo:
    Z_umap2 = umap_model.fit_transform(X)
    print("UMAP completado.")
except Exception as e:
    print("UMAP falló por incompatibilidad de versión sklearn/umap-learn.")
    print("Detalle del error:", repr(e))
    umap_error = True

# -----------------------------
# 5. Métrica de calidad: Silhouette con KMeans (k=10)
#    k=10 porque MNIST tiene dígitos 0..9
# -----------------------------

embeddings_dict = {
    "PCA 2D": Z_pca2,
    "Kernel PCA (RBF) 2D": Z_kpca2,
    "t-SNE 2D": Z_tsne2,
}

# Sólo agregamos UMAP si se logró calcular
if Z_umap2 is not None:
    embeddings_dict["UMAP 2D"] = Z_umap2

silhouette_scores = {}
for name, Z in embeddings_dict.items():
    # Clustering en el espacio reducido
    kmeans = KMeans(n_clusters=10, n_init=10, random_state=42)
    cluster_labels = kmeans.fit_predict(Z)

    # Silhouette mide qué tan bien separados quedaron los grupos
    sil = silhouette_score(Z, cluster_labels)
    silhouette_scores[name] = sil
    print(f"{name}: Silhouette = {sil:.4f}")

# -----------------------------
# 6. Comparación visual de las proyecciones
# -----------------------------
num_plots = len(embeddings_dict)
cols = 2
rows = int(np.ceil(num_plots / cols))

plt.figure(figsize=(12, 5*rows))

for i, (name, Z) in enumerate(embeddings_dict.items(), 1):
    plt.subplot(rows, cols, i)
    plt.scatter(Z[:, 0], Z[:, 1], c=y, cmap='tab10', s=5)
    plt.title(name)
    plt.xlabel("Comp 1")
    plt.ylabel("Comp 2")

plt.tight_layout()
plt.show()

# -----------------------------
# 7. Mostrar el resumen de Silhouette
# -----------------------------
print("\nResumen Silhouette por método:")
for name, val in silhouette_scores.items():
    print(f"{name:20s} -> {val:.4f}")

if umap_error:
    print("\nNota: UMAP no se evaluó por incompatibilidad entre esta versión de umap-learn y scikit-learn.")
```
