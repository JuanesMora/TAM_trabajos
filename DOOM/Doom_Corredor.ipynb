```python
import numpy as np
import gymnasium as gym
from gymnasium import spaces
import cv2 
import vizdoom 
from vizdoom import DoomGame, ScreenFormat, ScreenResolution, GameVariable
import os
import torch
from stable_baselines3 import PPO
from stable_baselines3.common.vec_env import DummyVecEnv

# --- 1. VERIFICACIONES ---
print(f"NumPy Version: {np.__version__}")
print(f"GPU Disponible: {torch.cuda.is_available()}")

# --- 2. CLASE ROBUST DOOM ENV ---
class RobustDoomEnv(gym.Env):
    metadata = {"render_modes": ["rgb_array"], "render_fps": 30}
    
    def __init__(self, config='deadly_corridor.cfg'): 
        super(RobustDoomEnv, self).__init__()
        self.game = DoomGame()
        
        # Carga segura de configuración
        scenarios_dir = os.path.join(os.path.dirname(vizdoom.__file__), 'scenarios')
        config_path = os.path.join(scenarios_dir, config)
        self.game.load_config(config_path)
        
        # Configuración
        self.game.set_doom_skill(1)  # Modo Fácil
        self.game.set_window_visible(False)
        self.game.set_screen_format(ScreenFormat.GRAY8) 
        self.game.set_screen_resolution(ScreenResolution.RES_640X480)
        
        # Variables para navegación
        self.game.add_available_game_variable(GameVariable.POSITION_X)
        self.game.add_available_game_variable(GameVariable.POSITION_Y)
        self.game.add_available_game_variable(GameVariable.HEALTH)
        self.game.init()
        
        # Espacios
        self.obs_shape = (100, 160, 1)
        self.observation_space = spaces.Box(low=0, high=255, shape=self.obs_shape, dtype=np.uint8)
        self.action_space = spaces.Discrete(7)
        self.available_buttons = self.game.get_available_buttons_size()
        
        # Meta del Deadly Corridor
        self.goal_pos = np.array([1100.0, 1000.0]) 
        self.last_distance = None 

    def _get_player_pos(self, state):
        if state is not None and state.game_variables is not None:
            try:
                return np.array([state.game_variables[0], state.game_variables[1]])
            except:
                return np.array([0.0, 0.0])
        return np.array([0.0, 0.0])
    
    def _get_distance(self, pos):
        return np.linalg.norm(pos - self.goal_pos)
        
    def preprocess(self, img):
        # BLINDAJE CONTRA ERRORES DE IMAGEN
        try:
            if img is None: 
                return np.zeros(self.obs_shape, dtype=np.uint8)
            img = np.array(img, dtype=np.uint8)
            img = cv2.resize(img, (160, 100))
            img = np.reshape(img, (100, 160, 1))
            return img
        except:
            return np.zeros(self.obs_shape, dtype=np.uint8)

    def step(self, action_index):
        buttons = [0] * self.available_buttons
        # Acciones: Strafe (1,2), Atacar (3), Correr (4), Girar (5,6)
        if action_index == 1 and self.available_buttons > 0: buttons[0] = 1 
        elif action_index == 2 and self.available_buttons > 1: buttons[1] = 1 
        elif action_index == 3 and self.available_buttons > 2: buttons[2] = 1 
        elif action_index == 4 and self.available_buttons > 3: buttons[3] = 1 
        elif action_index == 5 and self.available_buttons > 4: buttons[4] = 1 
        elif action_index == 6 and self.available_buttons > 5: buttons[5] = 1 
        
        base_reward = self.game.make_action(buttons, 4)
        current_state = self.game.get_state()
        done = self.game.is_episode_finished()
        
        total_reward = base_reward
        
        # Lógica de Navegación
        if current_state is not None:
            player_pos = self._get_player_pos(current_state)
            current_distance = self._get_distance(player_pos)
            if self.last_distance is not None:
                # Recompensa por acercarse a la meta
                total_reward += (self.last_distance - current_distance) * 0.05 
            self.last_distance = current_distance
            state_img = self.preprocess(current_state.screen_buffer)
        else:
            state_img = np.zeros(self.obs_shape, dtype=np.uint8)
        
        # --- CORRECCIÓN DE TIPOS ---
        return state_img, float(total_reward), bool(done), False, {}

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        self.game.new_episode()
        current_state = self.game.get_state()
        if current_state is not None:
            player_pos = self._get_player_pos(current_state)
            self.last_distance = self._get_distance(player_pos)
            state_img = self.preprocess(current_state.screen_buffer)
        else:
            self.last_distance = None
            state_img = np.zeros(self.obs_shape, dtype=np.uint8)
        return state_img, {}
    
    def close(self):
        self.game.close()


# --- 3. EJECUCIÓN DEL ENTRENAMIENTO ---
env = DummyVecEnv([lambda: RobustDoomEnv(config='deadly_corridor.cfg')])

print("\n🚀 Configurando PPO...")
model = PPO(
    "CnnPolicy", 
    env, 
    verbose=1, 
    learning_rate=0.0001, 
    n_steps=2048, 
    batch_size=64,
    gamma=0.99,
    device="cuda"  # Usar la GPU
)

print("🔥 Iniciando entrenamiento del Agente Corredor...")
model.learn(total_timesteps=150000) 
model.save("ppo_deadly_corridor_pro")
print("✅ Entrenamiento finalizado y modelo guardado.")
```

```python
import imageio
import numpy as np
from stable_baselines3 import PPO
from stable_baselines3.common.vec_env import DummyVecEnv
from IPython.display import display, HTML

# --- CONFIGURACIÓN ---
MODELO_A_CARGAR = "ppo_deadly_corridor_pro"
NOMBRE_VIDEO = "video_progreso_150k.gif"
MAX_STEPS_VIDEO = 1000 # Límite para que no sea eterno

print(f"🎥 Cargando modelo: {MODELO_A_CARGAR}...")

# 1. Crear entorno de evaluación
# Usamos la misma clase RobustDoomEnv que ya definiste
try:
    env_eval = DummyVecEnv([lambda: RobustDoomEnv(config='deadly_corridor.cfg')])
    
    # 2. Cargar el modelo entrenado
    # Importante: Usamos device='cuda' si entrenaste con GPU, o 'cpu' si prefieres.
    # SB3 suele manejarlo automático, pero lo forzamos a lo que usaste.
    model = PPO.load(MODELO_A_CARGAR, env=env_eval)
    print("✅ Modelo cargado correctamente.")
except Exception as e:
    print(f"❌ Error al cargar: {e}")
    # Si falla, detiene la ejecución aquí
    raise e

print("🎬 Grabando partida...")
frames = []
obs = env_eval.reset()
done = False
step = 0

# 3. Bucle de Grabación
while not done and step < MAX_STEPS_VIDEO:
    # Usamos deterministic=True para ver la MEJOR estrategia que aprendió (no exploración)
    action, _ = model.predict(obs, deterministic=True)
    
    obs, _, dones, _ = env_eval.step(action)
    done = dones[0]
    
    # Capturar frame
    # El entorno devuelve (1, 100, 160, 1) -> Tomamos el primer elemento y convertimos a uint8
    frame = obs[0].astype(np.uint8)
    
    # TRUCO VISUAL: Convertir Escala de Grises (1 canal) a RGB Falso (3 canales)
    # Esto es necesario para que los reproductores de video/GIF lo entiendan bien
    if frame.shape[-1] == 1:
        frame = np.repeat(frame, 3, axis=-1)
        
    frames.append(frame)
    step += 1
    
    if step % 100 == 0:
        print(f"   ... procesando frame {step}")

env_eval.close()

# 4. Guardar y Mostrar
print(f"💾 Guardando GIF ({len(frames)} frames)...")
imageio.mimsave(NOMBRE_VIDEO, frames, fps=25)

print(f"✅ Video '{NOMBRE_VIDEO}' generado.")
print("👇 Aquí tienes el resultado de tu entrenamiento de 150k pasos:")

# Mostrar en el notebook
display(HTML(f'<img src="{NOMBRE_VIDEO}" controls autoplay loop width="640">'))
```
